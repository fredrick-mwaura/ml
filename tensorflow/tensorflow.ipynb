# Creating random tensors
*Random tensors and tensors of some arbitrary size which contain random number*

```python
# >>>when 2 seeds are set (global & operational), they are used in conjuction by tf to determine the the random sequence
# >>>if global and operational level random seed are same no shuffling is made
tf.random.set_seed(90)# global random seed
rand_1 = tf.random.Generator.from_seed(42)
rand_1 = rand_1.normal(shape=(3,2))
rand_2 = tf.random.Generator.from_seed(12)
rand_2 = rand_2.normal(shape=(3, 2))

rand_1, rand_2, rand_1 == rand_2
```

# shuffle images in a tensor

```python
not_shuf = tf.constant([[10, 2],
                        [30, 2],
                        [3, 5]])
#shuffling
shuf = tf.random.shuffle(not_shuf, seed=42)
shuf
```

# generators from seed

```python
g = tf.random.Generator.from_seed(42)
g.normal(shape=(2,2)).numpy()
```

```python
#from non-deterministic state
nG = tf.random.Generator.from_non_deterministic_state()
nG = nG.normal(shape=(2,3))
```

```python
tf.random.shuffle(nG)
```

**multiplying matrices**

*must have the similar inner axis*

```python
x = tf.constant([[1, 2, 1],
                 [0, 1,0],
                 [2, 3, 4]])
y = tf.constant([[2, 5],
                 [6, 7],
                 [1, 8]])
x @ y
```

**Dot Product**

* tf.linalg.matmul()

* tf.tensordot()

* **when multiplying two matrix use .transponse instead of reshaping the tensor**

```python
# x, y requires to be transposed
tf.tensordot(tf.transpose(x), y, axes=1) # tf.linalg.tensordot() -> main alias
```

```python
##matrix multiplication btwn x and y (transposed)
tf.matmul(x, tf.transpose(y))
```

```python
tf.matmul(x, tf.reshape(y, (3, 2)))
y
# tf.matmul(x, tf.transpose(y))
```
