# Creating random tensors
*Random tensors and tensors of some arbitrary size which contain random number*

```python
# >>>when 2 seeds are set (global & operational), they are used in conjuction by tf to determine the the random sequence
# >>>if global and operational level random seed are same no shuffling is made
tf.random.set_seed(90)# global random seed
rand_1 = tf.random.Generator.from_seed(42)
rand_1 = rand_1.normal(shape=(3,2))
rand_2 = tf.random.Generator.from_seed(12)
rand_2 = rand_2.normal(shape=(3, 2))

rand_1, rand_2, rand_1 == rand_2
```

# Positional maximum and minimum
* in which maximum and/or minimum occurs

```python
tf.random.set_seed(40)
f = tf.random.uniform(shape=[50])
f
```

```python
## return the largest value across axes of a tensor
tf.argmax(f)
```

```python
#index on the largest value
ne = f[tf.argmax(f)]
```

```python
#using max method
tf.reduce_max(f)
```

```python
assert tf.reduce_max(f) == f[tf.argmax(f)]
```

```python
#positional minimum
tf.argmin(f)
f[tf.argmin(f)]
```

```python
tf.reduce_min(f)
```

```python
assert tf.reduce_min(f) == f[tf.argmin(f)]
```

# One hot encoding

```python
depth = 4
fruits = [1, 3, 9, 6]
tf.one_hot(fruits, depth, on_value=5.0, off_value=-1.0) # categorical indices to binary matrrix presentations
```

```python
#math ops with one hot encoding
h = tf.range(1, 10)
h = tf.square(h)
tf.math.sqrt(tf.cast(h, dtype=tf.float32))
```

```python
##logging
tf.math.log(tf.cast(h, dtype=tf.float32))
# tf.math.cos(tf.cast(h, dtype=tf.float32))
```
